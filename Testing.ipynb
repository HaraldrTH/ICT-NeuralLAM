```python
import os
import numpy as np
test_data_loc = "Test_Data"
test_data = []
for data_file in os.listdir(test_data_loc):
    if data_file.endswith(".npy"):
        data = np.load(os.path.join(test_data_loc, data_file))
        test_data.append(data)
        print(f"Loaded {data_file} with shape {data.shape}")
```
